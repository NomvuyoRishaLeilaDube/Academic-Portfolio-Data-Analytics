## Why did I choose Logistic Regression (LR)

### What Is LR?
Although the name suggests a regression method, LR is in fact a classification algorithm that predicts the probability of a data point belonging to a particular class (Chakraborty, 2025). It does this using the sigmoid function, which maps any input value to a probability score between 0 and 1. Predictions are then made by applying a decision threshold (usually 0.5), so that cases with a probability at or above the threshold are assigned to that class, and those below are not (Eftekharian et al., 2025).

### Now Let Me Answer The Golden Question: Why Did I Choose It?

1) In a multiclass setting such as this one My chosen dataset presentsa multicall of variables, whereby there are five possible cancer types (Masryo, 2025). LR uses a one-vs-rest strategy, training a separate binary classifier for each class and selecting the class with the highest predicted probability (GeeksforGeeks, 2025)

2) Since this is dataset is set in a medical context LR is suitable because it is transparent. Each feature in the model has a corresponding coefficient that shows exactly how much it increases or decreases the probability of a given cancer type (GeeksforGeeks, 2025). This type of interpretability would be important in instances whereby scheme needs to explain or audit a decision, thus this allows for the regulators and clinicians can understand the reasoning behind each prediction (Chakraborty, 2025).

3) Instead of just returning a class label, LR outputs provide a probability score (Eftekharian et al., 2025).. In the context of insurance benefit claims, this would be valuable whereby the scheme can use the probability to flag borderline cases for human review rather than applying a purely binary decision (Eftekharian et al., 2025).

4) Additionally, LR has a strong track record in medical diagnostic tasks. In a study evaluating it on the Wisconsin Diagnostic Breast Cancer dataset across 50 independent runs, the model achieved a mean accuracy of 93.8%, a sensitivity of 97.2% for detecting malignant cases, and an AUROC of 0.9452 (Eftekharian et al., 2025). Moreover, for a 2022 cervical cancer study, the LR was able to predict cervical cancer and was shown to produce fairly reasonable results even with smaller datasets (Krawi, 2022).

5) Looking beyond this assignment, LR is generally computationally lightweight, which is important when a scheme needs to process many claims quickly. In addition, the models are quick to train with LR, and we can get prediction essentally in real-time. Thus LR is an algorithm which can be beneficial for  processing pipelines (Chakraborty, 2025). Now back to my assignment, LR is lightweight so that's good for me, training is fast which is another bonus, plus I'll get me prediction in real-time which is perfect:)

# THEORY


## What is Logstic Regression

Logistic Regression (LR) is a machine learning classification algorithm that uses statistical methods to evaluate the relationships between variables (Krawi, 2022). As contradictory as it name might seem, LR it is not a regression algorithm, but rather it is a classification algorithm which is used to predict binary outcomes, such as whether a tumour is malignant or benign (Chakraborty, 2025). It works by using the sigmoid function to squash predictions into a probability score between 0 and 1, and predictions are then made by applying a decision threshold of 0.5 (Eftekharian et al., 2025). Follwing this, instances whereby the probabilities are at or above the threshold are assigned to that class, and those below are not (Eftekharian et al., 2025). An advantage of LR is that it does not require large amounts of data in order to get reasonable results (Krawi, 2022).

## What is Naïve Bayes

Naïve Bayes (NB) is a probabilistic machine learning classification algorithm whose function is to use probability to predict the category of a data point (GeeksforGeeks, 2026). It is based on Bayes' Theorem and assumes that all features are independent of one another (GeeksforGeeks, 2026).This means that the presence of one feature does not affect any other feature, they are indepdent from one another. This assumption of indepdence is why this algorithm is reffered to as "Naïve" (GeeksforGeeks, 2026). NB is able to produce predictions at a faster speed than other classifiers due to the fact that it has very few parameters (GeeksforGeeks, 2026).

## Assumptions of Naïve Bayes (GeeksforGeeks, 2026)
Before using a NB model, five key assumptions needs to be checked:

1) Features are independent => each feature contributes independetly to the predictions with no relation to any other feature.

2) Normality of continuous features => within each class, continuous features need to be normally distributed.

3) Categorical or count-based features have multinomial distributions => the algorithm assumes that its values are spread across multiple possible outcomes in a way that follows a multinomial distribution within each class

4) All features are of equal importance => the algorithm treats every feature as having the same level of influence on the final prediction, whereby no single feature is considered more important than another when determining which class a data point belongs to

5) The data should not contain any missing values

## Types of Naïve Bayes (GeeksforGeeks, 2026)

1) Gaussian Naïve Bayes (GNB) => this variant is applied when the features in the dataset are continuous and are expected to be normally distributed within each class. I chose this algorithm for this assignment because my dataset includes continuous variables such as Age, BMI, and the overall risk score.

2) Multinomial Naïve Bayes => this version is suited to situations where features capture how often something occurs, for example how many times a word appears in a piece of text. It's mostly used where there are tasks involving document or text classification.

3) Bernoulli Naïve Bayes => rather than counting how often a feature appears, this variant is concerned only with whether a feature is present or not. It works best in cases where a simple yes or no distinction between features carries more meaning than the actual count.

## Examples of Where Naïve Bayes Has Been Applied

1) Medical diagnosis => it has been used to estimate how likely a patient is to have a particular condition given a set of observed symptoms (GeeksforGeeks, 2026).

2) Breast cancer classification => on the Wisconsin Breast Cancer Dataset, the model reached an accuracy of 97.08% while also achieving perfect specificity, meaning it did not produce a single false positive for benign cases (Zhang et al., 2024).

3) Multi-cancer prediction => when the Gaussian variant was applied across both breast and lung cancer datasets, it returned accuracy rates of 98% and 90% respectively (Banu & Thirumalaikolundusubramanian, 2018).

4) Credit scoring => the algorithm has also been used in finance, whereby it can help assess whether a person is likely to repay the money they borrowed(GeeksforGeeks, 2026).

### Identify outliers using the IQR

The boxplots helped me identify the presence of outliers in a few variables, however I did not do this for all 19 features. Thus, I will now statistically identify the outliers, then determine if I should remove them.



LET ME PROVIDE SOME CONTEXT TO THE PROCESSES THAT WILL SUBSEQUENTLY FOLLOW:

To try and measure the datas distribution, we us the IQR. It particularly is used to measures how spread out the middle 50% of the data is. It is calculated as (Patil, 2023):

IQR = Q3 − Q1 (Patil, 2023)

To find outliers, two limits are set:

Lower bound: Q1 − 1.5 × IQR (Patil, 2023) Upper bound: Q3 + 1.5 × IQR (Patil, 2023)

Values that are identified bove the upper bound or below the lower bound are outliers (Patil, 2023). This means thats these values are much smaller or larger than most of the data, their essentially anomolies.

I will only be indentifying, then removing the outliers for the continuous numerical variables, which are:

*   Age
*   BMI
*   Overall_Risk_Score



```python
# Finding the number of Age outliers

q1 = df['Age'].quantile(0.25)
q3 = df['Age'].quantile(0.75)
IQR = q3 - q1

lower = q1 - 1.5*IQR
upper = q3 + 1.5*IQR

outliers = df[(df['Age'] < lower) | (df["Age"] > upper)]

print(len(outliers))
```

```python
# Finding the number of BMI outliers

q1 = df['BMI'].quantile(0.25)
q3 = df['BMI'].quantile(0.75)
IQR = q3 - q1

lower = q1 - 1.5*IQR
upper = q3 + 1.5*IQR

outliers = df[(df['BMI'] < lower) | (df["BMI"] > upper)]

print(len(outliers))
```

Now that we have visually determined normaility, let's do it statistically using the Shapiro-Wilks Test

Normality can be checked visually using a histrogram or statistically using the Shapiro-Wilk test (P>0.05 normality assumption is met) (Roustaei, 2024).

```python
# attributed: whole code
#author (Nadia, 2026)

from scipy.stats import shapiro

for col in df.select_dtypes(include='number').columns:
    stat, p = shapiro(df[col].dropna())
    if p > 0.05:
        print(f"{col} looks normally distributed (p={p:.3f})\n")
    else:
        print(f"{col} is NOT normally distributed (p={p:.3f})\n")
```

So upon statisticasl analysis, none of the data is normally distributed. Interesting.

So to check for correlations (or lack thereof) we can use using the Variance Inflation Factro (VIF), whereby a VIF of 1 indicates a lack of correlation, between 1-5 is a mild correlation, between 5-10 a strong correlation while a VIF of above 10 is now a serious problem to the point that predictions would now become reliable (Roustaei, 2024)

```python
#VIF SCORE

# measures multicollinearity, indicating how much a regression coefficient's variance is inflated by correlated predictors.
# A VIF of 1 indicates no correlation,
# while values or suggest high, problematic multicollinearity requiring correction. Low VIFs (near 1) mean predictors are independent, enhancing model stability

# attributed = whole code
# author (Sarina, 2026)

from statsmodels.tools.tools import add_constant
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
# attributed = whole code
# author (Sarina, 2026)

def compute_vif(df):
  X = add_constant(df)

  vif = pd.DataFrame()
  vif['feature'] = X.columns
  vif["VIF"] = [variance_inflation_factor (X.values, i)
                for i in range(X.shape[1])]

  return(vif)
```

```python
#USAGE

X = df.drop(columns=['Cancer_Type'])
vif_df = compute_vif(X)
print(vif_df.sort_values(by='VIF', ascending=False))
```
